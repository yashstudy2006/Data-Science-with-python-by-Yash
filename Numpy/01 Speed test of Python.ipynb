```python
# using python list
import numpy as np
import time
size = 1_000
start = time.time()
l1 = list(range(size))
l2 = list(range(size))
add = [x + y for x,y in zip(l1,l2)]
end = time.time()
print(add[0:10])
print(end - start)
```

```python
# using Numpy Array
import numpy as np
import time
size = 1_000
start = time.time()
l1 = np.array(list(range(size)))
l2 = np.array(list(range(size)))
add = l1 + l2
end = time.time()
print(add[0:10])
print(end - start)
```
